```python
import random
from datetime import datetime, timedelta
from pathlib import Path

import pandas as pd


random.seed(42)

BASE_DIR = Path(__file__).resolve().parent.parent
DATA_DIR = BASE_DIR / "data"

DATA_DIR.mkdir(parents=True, exist_ok=True)


categories = [
    {"category_id": 1, "category_name": "Electronics"},
    {"category_id": 2, "category_name": "Home"},
    {"category_id": 3, "category_name": "Sports"},
    {"category_id": 4, "category_name": "Accessories"},
]

df_categories = pd.DataFrame(categories)


products = [
    ("Wireless Headphones", 1, 180000, 45),
    ("Bluetooth Speaker", 1, 220000, 30),
    ("Smart Watch", 1, 450000, 20),
    ("USB-C Charger", 1, 85000, 70),
    ("Coffee Maker", 2, 280000, 25),
    ("Desk Lamp", 2, 95000, 40),
    ("Air Fryer", 2, 390000, 18),
    ("Electric Kettle", 2, 150000, 35),
    ("Yoga Mat", 3, 75000, 60),
    ("Dumbbell Set", 3, 210000, 25),
    ("Running Shoes", 3, 320000, 30),
    ("Sports Backpack", 3, 170000, 40),
    ("Leather Wallet", 4, 120000, 50),
    ("Sunglasses", 4, 140000, 35),
    ("Travel Organizer", 4, 90000, 45),
]

df_products = pd.DataFrame(
    [
        {
            "product_id": i + 1,
            "product_name": product[0],
            "category_id": product[1],
            "price": product[2],
            "stock": product[3],
        }
        for i, product in enumerate(products)
    ]
)


first_names = [
    "Juan", "Carlos", "Andres", "Daniel", "Sebastian",
    "Mateo", "Nicolas", "Santiago", "Felipe", "David",
    "Laura", "Camila", "Valentina", "Maria", "Sofia",
    "Paula", "Daniela", "Natalia", "Carolina", "Alejandra",
]

last_names = [
    "Rodriguez", "Gomez", "Rojas", "Martinez", "Lopez",
    "Garcia", "Hernandez", "Perez", "Torres", "Ramirez",
    "Castro", "Moreno", "Vargas", "Jimenez", "Diaz",
]

cities = [
    "Bogota",
    "Medellin",
    "Cali",
    "Barranquilla",
    "Cartagena",
    "Bucaramanga",
    "Pereira",
    "Manizales",
]

customers = []

start_date = datetime(2025, 9, 23)
end_date = datetime(2026, 9, 23)

for customer_id in range(1, 31):
    first_name = random.choice(first_names)
    last_name = random.choice(last_names)

    registration_date = start_date + timedelta(
        days=random.randint(0, 365)
    )

    customers.append(
        {
            "customer_id": customer_id,
            "first_name": first_name,
            "last_name": last_name,
            "email": f"{first_name.lower()}.{last_name.lower()}{customer_id}@example.com",
            "city": random.choice(cities),
            "registration_date": registration_date.date(),
        }
    )

df_customers = pd.DataFrame(customers)


statuses = [
    "Completed",
    "Completed",
    "Completed",
    "Completed",
    "Pending",
    "Cancelled",
]

payment_methods = [
    "Credit Card",
    "Debit Card",
    "PSE",
    "Cash",
    "Digital Wallet",
]

orders = []

for order_id in range(1, 101):
    customer_id = random.randint(1, 30)

    order_date = start_date + timedelta(
        days=random.randint(0, 365)
    )

    orders.append(
        {
            "order_id": order_id,
            "customer_id": customer_id,
            "order_date": order_date.date(),
            "status": random.choice(statuses),
            "payment_method": random.choice(payment_methods),
        }
    )

df_orders = pd.DataFrame(orders)


order_details = []

for order_id in range(1, 101):
    num_products = random.choice([2, 2, 3])

    selected_products = random.sample(
        range(1, 16),
        num_products
    )

    for product_id in selected_products:
        product = df_products[
            df_products["product_id"] == product_id
        ].iloc[0]

        order_details.append(
            {
                "order_id": order_id,
                "product_id": product_id,
                "quantity": random.randint(1, 4),
                "unit_price": product["price"],
            }
        )

df_order_details = pd.DataFrame(order_details)

if len(df_order_details) > 250:
    df_order_details = df_order_details.iloc[:250].copy()

elif len(df_order_details) < 250:
    existing_pairs = set(
        zip(
            df_order_details["order_id"],
            df_order_details["product_id"],
        )
    )

    while len(df_order_details) < 250:
        order_id = random.randint(1, 100)
        product_id = random.randint(1, 15)

        if (order_id, product_id) in existing_pairs:
            continue

        product = df_products[
            df_products["product_id"] == product_id
        ].iloc[0]

        new_detail = {
            "order_id": order_id,
            "product_id": product_id,
            "quantity": random.randint(1, 4),
            "unit_price": product["price"],
        }

        df_order_details = pd.concat(
            [
                df_order_details,
                pd.DataFrame([new_detail]),
            ],
            ignore_index=True,
        )

        existing_pairs.add((order_id, product_id))

df_order_details.insert(
    0,
    "order_detail_id",
    range(1, len(df_order_details) + 1)
)


df_categories.to_csv(
    DATA_DIR / "categories.csv",
    index=False
)

df_products.to_csv(
    DATA_DIR / "products.csv",
    index=False
)

df_customers.to_csv(
    DATA_DIR / "customers.csv",
    index=False
)

df_orders.to_csv(
    DATA_DIR / "orders.csv",
    index=False
)

df_order_details.to_csv(
    DATA_DIR / "order_details.csv",
    index=False
)


print(f"Categorías: {len(df_categories)}")
print(f"Productos: {len(df_products)}")
print(f"Clientes: {len(df_customers)}")
print(f"Órdenes: {len(df_orders)}")
print(f"Detalles: {len(df_order_details)}")
print(f"Archivos guardados en: {DATA_DIR}")
```
